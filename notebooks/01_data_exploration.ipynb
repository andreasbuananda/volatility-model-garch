# 01 — Exploratory Data Analysis (EDA)

This notebook downloads BTC-USD daily OHLCV data from Yahoo Finance,
calculates log returns, and performs initial exploratory visualizations.

**Pipeline step 1 of 7.**

```python
import sys
sys.path.insert(0, "..")

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

from src.data_loader import prepare_dataset
from src.preprocessing import compute_log_returns

sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1.1 Download and Prepare Data

```python
df = prepare_dataset(start_date="2017-01-01")
print(df.shape)
df.head()
```

## 1.2 BTC Closing Price

```python
fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(df.index, df["Close"], linewidth=0.8, color="steelblue")
ax.set_title("BTC-USD Daily Closing Price")
ax.set_xlabel("Date")
ax.set_ylabel("Price (USD)")
plt.tight_layout()
plt.savefig("../results/figures/btc_price.png", dpi=150)
plt.show()
```

## 1.3 Log Returns

```python
returns = df["log_return"]

fig, axes = plt.subplots(2, 1, figsize=(14, 7))
axes[0].plot(returns.index, returns.values, linewidth=0.6, color="steelblue")
axes[0].set_title("BTC-USD Log Returns")
axes[0].set_ylabel("Log Return")

axes[1].hist(returns, bins=100, color="steelblue", alpha=0.7, edgecolor="white")
axes[1].set_title("Distribution of Log Returns")
axes[1].set_xlabel("Log Return")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.savefig("../results/figures/btc_returns.png", dpi=150)
plt.show()
```

## 1.4 Descriptive Statistics

```python
desc = returns.describe()
print(desc)
print(f"
Skewness : {returns.skew():.4f}")
print(f"Kurtosis : {returns.kurtosis():.4f}")
```

## 1.5 Volatility Clustering

Squared returns serve as a proxy for realized variance. Strong autocorrelation
in squared returns is a hallmark of volatility clustering.

```python
sq_returns = returns ** 2
fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(sq_returns.index, sq_returns.values, linewidth=0.6, color="coral")
ax.set_title("Squared Log Returns (Volatility Proxy)")
ax.set_xlabel("Date")
ax.set_ylabel("Squared Return")
plt.tight_layout()
plt.savefig("../results/figures/btc_sq_returns.png", dpi=150)
plt.show()
```
