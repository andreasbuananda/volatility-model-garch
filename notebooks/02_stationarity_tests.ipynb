# 02 — Stationarity Tests (ADF & KPSS)

This notebook tests whether BTC log returns are stationary using the
Augmented Dickey-Fuller (ADF) and KPSS tests. Stationarity is a prerequisite
for ARMA and GARCH modeling.

**Pipeline step 2 of 7.**

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.data_loader import load_processed_data
from src.preprocessing import adf_test, kpss_test

%matplotlib inline
```

## 2.1 Load Processed Returns

```python
df = load_processed_data()
returns = df["log_return"].dropna()
print(f"Sample size: {len(returns)}")
returns.describe()
```

## 2.2 ADF Test on Log Returns

H0: Series has a unit root (non-stationary)
H1: Series is stationary

For log returns, we expect to reject H0.

```python
adf_stat, adf_pval, is_stationary = adf_test(returns, significance=0.05, verbose=True)
```

## 2.3 KPSS Test on Log Returns

H0: Series is stationary
H1: Series has a unit root

For log returns, we expect to fail to reject H0.

```python
kpss_stat, kpss_pval, kpss_stationary = kpss_test(returns, significance=0.05, verbose=True)
```

## 2.4 Test on Raw Prices (for comparison)

Raw prices are expected to be non-stationary (unit root).

```python
prices = df["Close"]
adf_price_stat, adf_price_pval, price_stationary = adf_test(prices, verbose=True)
print(f"
Is BTC price stationary? {price_stationary}")
```
